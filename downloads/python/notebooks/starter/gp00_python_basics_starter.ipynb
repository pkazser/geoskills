## 10. Πρώτη επαφή με `GeoPandas`

Σε αυτό το σημείο δεν θα κάνουμε ακόμα ανάλυση. Θέλουμε απλώς να δούμε πώς από X/Y μπορούμε να δημιουργήσουμε γεωμετρίες σημείων.


```python
import geopandas as gpd
from shapely.geometry import Point

data = {
    "name": ["A", "B", "C"],
    "longitude": [23.55, 23.56, 23.57],
    "latitude": [41.09, 41.10, 41.08]
}

geometry = [Point(xy) for xy in zip(data["longitude"], data["latitude"])]

gdf = gpd.GeoDataFrame(data, geometry=geometry, crs="EPSG:4326")
gdf
```
